## Exercice 1 — Factorielle et puissance  ⭐

### Question 1

#### 📝 ÉNONCÉ

> Sans regarder le cours, écrire une fonction récursive `factorielle(n)` qui renvoie $n! = 1 \times 2 \times \cdots \times n$, avec $0! = 1$.

#### ✅ RÉSOLUTION

````python
def factorielle(n):
    if n == 0:
        return 1
    return n * factorielle(n - 1)
````

*Les tests de la feuille d'exercices :*

````python
assert factorielle(0) == 1
assert factorielle(1) == 1
assert factorielle(5) == 120
assert factorielle(10) == 3628800
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Cas de base :** `factorielle(0)` vaut `1`.
- **Cas récursif :** $n! = n \times (n-1)!$.
- **Erreur fréquente :** prendre `return 0` comme cas de base : toutes les multiplications donneraient alors `0`.

### Question 2

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `puissance(x, n)` qui renvoie $x^n$ (pour `n` entier positif ou nul), **sans** utiliser `**`.
>
> *Indice : $x^0 = 1$ et $x^n = x \times x^{n-1}$.*

#### ✅ RÉSOLUTION

````python
def puissance(x, n):
    if n == 0:
        return 1
    return x * puissance(x, n - 1)
````

*Les tests de la feuille d'exercices :*

````python
assert puissance(2, 0) == 1
assert puissance(2, 10) == 1024
assert puissance(5, 3) == 125
assert puissance(-3, 3) == -27
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Même schéma que la factorielle : on multiplie par `x` à chaque appel, et on s'arrête à `n == 0` où la réponse est `1`.
- Cette version fait `n` multiplications. On peut faire **beaucoup** mieux avec l'idée $x^n = x^{n/2} \times x^{n/2}$ quand `n` est pair : c'est l'**exponentiation rapide**, un des premiers exemples de « diviser pour régner » (pour $n = 100$ : 8 multiplications au lieu de 100).

## Exercice 2 — Maximum et nombre d'occurrences  ⭐⭐

### Question 1

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `maximum(t, i)` qui renvoie le plus grand élément de la liste non vide `t` à partir de l'indice `i`. On l'appelle avec `maximum(t, 0)` pour toute la liste.
>
> *Indice : quel est le cas de base, quand il ne reste qu'un seul élément ?*

#### ✅ RÉSOLUTION

````python
def maximum(t, i):
    if i == len(t) - 1:             # il ne reste qu'un élément : le dernier
        return t[i]
    m = maximum(t, i + 1)           # le plus grand de la suite de la liste
    if t[i] > m:
        return t[i]
    return m
````

*Les tests de la feuille d'exercices :*

````python
assert maximum([3, 8, 2, 5], 0) == 8
assert maximum([7], 0) == 7
assert maximum([-5, -2, -9], 0) == -2
assert maximum([1, 2, 3, 4], 0) == 4
assert maximum([3, 8, 2, 5], 2) == 5          # à partir de l'indice 2 : [2, 5]
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Cas de base :** quand `i` est l'indice du **dernier** élément, il ne reste que lui : c'est le maximum. (On ne peut pas prendre `i == len(t)` comme pour la somme : le maximum d'une liste vide n'a pas de sens.)
- **Cas récursif :** on fait confiance à l'appel récursif pour donner le maximum `m` de la suite de la liste, puis on le compare à `t[i]`.
- On range le résultat dans une variable `m` pour ne faire l'appel récursif **qu'une seule fois**. Écrire `if t[i] > maximum(t, i + 1): ... else: return maximum(t, i + 1)` le ferait deux fois, et le nombre d'appels doublerait à chaque niveau !

### Question 2

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `compte(t, x, i)` qui renvoie le nombre d'éléments égaux à `x` dans la liste `t` à partir de l'indice `i`.

#### ✅ RÉSOLUTION

````python
def compte(t, x, i):
    if i == len(t):
        return 0
    if t[i] == x:
        return 1 + compte(t, x, i + 1)
    return compte(t, x, i + 1)
````

*Les tests de la feuille d'exercices :*

````python
assert compte([1, 2, 1, 1, 3], 1, 0) == 3
assert compte([1, 2, 3], 5, 0) == 0
assert compte([], 5, 0) == 0
assert compte(["a", "b", "a"], "a", 0) == 2
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Même schéma que `somme_liste` du cours : cas de base `i == len(t)` (plus rien à compter), puis on traite `t[i]` et on laisse l'appel récursif compter dans la suite.
- `t[i]` compte pour `1` s'il est égal à `x`, pour `0` sinon.

## Exercice 3 — Les tours de Hanoï  ⭐⭐⭐

On a trois piquets (numérotés 1, 2, 3) et `n` disques de tailles différentes, empilés sur le piquet 1 du plus grand (en bas) au plus petit (en haut). Il faut tout déplacer sur le piquet 3, en respectant deux règles :
- on ne déplace **qu'un disque à la fois** (celui du dessus d'une pile) ;
- on ne pose **jamais** un disque sur un disque plus petit.

**L'idée récursive :** pour déplacer `n` disques du piquet `depart` au piquet `arrivee` :
1. on déplace les `n - 1` disques du dessus sur le **troisième** piquet (récursivement) ;
2. on déplace le plus grand disque sur `arrivee` ;
3. on redéplace les `n - 1` disques du troisième piquet sur `arrivee` (récursivement).

### Question 1

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `hanoi(n, depart, arrivee, autre)` qui **affiche** les déplacements à faire (par exemple `1 -> 3`) et **renvoie** le nombre de déplacements effectués.

#### ✅ RÉSOLUTION

````python
def hanoi(n, depart, arrivee, autre):
    if n == 0:
        return 0                                     # aucun disque : rien à faire
    nb1 = hanoi(n - 1, depart, autre, arrivee)       # 1. les n - 1 disques du dessus sur "autre"
    print(depart, "->", arrivee)                     # 2. le plus grand disque
    nb2 = hanoi(n - 1, autre, arrivee, depart)       # 3. les n - 1 disques sur "arrivee"
    return nb1 + 1 + nb2
````

*Les tests de la feuille d'exercices :*

````python
print("Déplacements pour 3 disques :")
assert hanoi(3, 1, 3, 2) == 7
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- La fonction traduit directement les 3 étapes. Le plus impressionnant : **on n'a pas besoin de savoir comment** déplacer les `n - 1` disques, on fait confiance à l'appel récursif.
- **Cas de base :** `n == 0`, il n'y a rien à faire. (On peut aussi prendre `n == 1` : un seul déplacement.)
- **Avec des boucles**, ce problème serait très difficile à écrire : il faudrait retenir soi-même tous les déplacements « en attente ». C'est un problème où la récursivité est vraiment nécessaire (astuce 1 : deux appels récursifs).
- Attention à l'**ordre des piquets** dans les appels : à l'étape 1, le piquet intermédiaire est `autre` et c'est `arrivee` qui sert de piquet libre.

### Question 2

#### 📝 ÉNONCÉ

> On note `d(n)` le nombre de déplacements pour `n` disques. D'après l'idée récursive, `d(n) = 2 × d(n - 1) + 1`, avec `d(0) = 0`.
>
> Écrire une fonction récursive `nb_deplacements(n)`, puis l'utiliser pour `n` de 1 à 10. Quelle formule devine-t-on ? Une légende raconte que des moines déplacent 64 disques d'or, un par seconde : combien de temps leur faut-il ?

#### ✅ RÉSOLUTION

````python
def nb_deplacements(n):
    if n == 0:
        return 0
    return 2 * nb_deplacements(n - 1) + 1


for n in range(1, 11):
    print(n, "disques :", nb_deplacements(n), "déplacements")

print()
secondes = nb_deplacements(64)
print("64 disques :", secondes, "secondes, soit environ", secondes // (3600 * 24 * 365), "ans")
````

*Les tests de la feuille d'exercices :*

````python
assert nb_deplacements(0) == 0
assert nb_deplacements(3) == 7
for n in range(1, 11):
    assert nb_deplacements(n) == 2 ** n - 1
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- On trouve `1, 3, 7, 15, 31, ...` : à chaque fois une puissance de 2 moins 1. La formule est $d(n) = 2^n - 1$.
- Pour 64 disques : $2^{64} - 1 \approx 1{,}8 \times 10^{19}$ secondes, soit environ **585 milliards d'années**, plus de 40 fois l'âge de l'Univers.
- C'est un coût **exponentiel** : ajouter un seul disque double le travail. On ne peut pas faire mieux pour les tours de Hanoï, mais c'est un ordre de grandeur à éviter en général.

## Exercice 4 — Remplir une zone dans une image  ⭐⭐⭐

📚 **Tombé au bac :** épreuve pratique 2024, sujet 27 : `colore_comp1(M, i, j, val)`, qui donne la valeur `val` à tous les pixels de la zone du pixel `(i, j)`.

Une image en noir et blanc est une matrice `M` dont les cases (les **pixels**) valent `0` (blanc) ou `1` (noir). Deux pixels noirs sont **voisins** s'ils se touchent par un côté (en haut, en bas, à gauche ou à droite). Une **zone** (on dit aussi une *composante connexe*) est un ensemble de pixels noirs reliés entre eux de voisin en voisin.

C'est l'outil « pot de peinture » des logiciels de dessin : on clique sur un pixel, et toute sa zone change de couleur.

La fonction `afficher` est donnée.

### Question 1

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `colorer(M, i, j, val)` qui donne la valeur `val` (un entier supérieur ou égal à 2) à tous les pixels de la zone qui contient le pixel `(i, j)`. Si `(i, j)` est en dehors de l'image, ou si ce n'est pas un pixel noir (`1`), la fonction ne fait rien. La fonction **modifie** `M` et ne renvoie rien.
>
> *Indice : on colore le pixel `(i, j)`, puis on appelle `colorer` sur ses 4 voisins.*

#### ✅ RÉSOLUTION

````python
def afficher(M):
    for ligne in M:
        for x in ligne:
            print(x, end=" ")
        print()
    print()


def colorer(M, i, j, val):
    if i < 0 or i >= len(M) or j < 0 or j >= len(M[0]):
        return                       # en dehors de l'image
    if M[i][j] != 1:
        return                       # pixel blanc, ou déjà coloré
    M[i][j] = val
    colorer(M, i - 1, j, val)        # en haut
    colorer(M, i + 1, j, val)        # en bas
    colorer(M, i, j - 1, val)        # à gauche
    colorer(M, i, j + 1, val)        # à droite
````

````python
M = [[1, 1, 0, 0, 1],
     [0, 1, 0, 1, 1],
     [0, 1, 0, 0, 0],
     [1, 0, 0, 1, 1]]
afficher(M)
colorer(M, 0, 0, 2)
afficher(M)
````

*Les tests de la feuille d'exercices :*

````python
M = [[1, 1, 0, 0, 1],
     [0, 1, 0, 1, 1],
     [0, 1, 0, 0, 0],
     [1, 0, 0, 1, 1]]
colorer(M, 0, 0, 2)
assert M == [[2, 2, 0, 0, 1],
             [0, 2, 0, 1, 1],
             [0, 2, 0, 0, 0],
             [1, 0, 0, 1, 1]]
colorer(M, 0, 2, 3)              # pixel blanc : rien ne change
colorer(M, 9, 9, 3)              # en dehors : rien ne change
assert M[0][2] == 0
colorer(M, 1, 4, 3)
assert M[0][4] == 3 and M[1][3] == 3 and M[1][4] == 3 and M[3][3] == 1
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Deux cas de base :** on est sorti de l'image, ou le pixel n'est pas noir. Dans les deux cas, on s'arrête sans rien faire (`return` seul).
- **Le point crucial :** on colore le pixel **avant** de lancer les appels sur les voisins. Comme un pixel coloré ne vaut plus `1`, on ne le traitera **pas deux fois**. Sinon, deux voisins s'appelleraient l'un l'autre à l'infini.
- C'est pour ça que `val` doit être différent de `1` (et de `0`) : on prend `2`, `3`...
- **Avec des boucles**, ce problème serait très difficile à écrire. C'est un exemple où la récursivité est vraiment l'outil naturel.

### Question 2

#### 📝 ÉNONCÉ

> En utilisant `colorer`, écrire une fonction `nombre_zones(M)` qui renvoie le nombre de zones noires de l'image. On pourra colorer chaque nouvelle zone rencontrée avec une valeur différente : `2`, puis `3`, puis `4`...

#### ✅ RÉSOLUTION

````python
def nombre_zones(M):
    val = 2
    for i in range(len(M)):
        for j in range(len(M[0])):
            if M[i][j] == 1:                 # pixel noir pas encore coloré : nouvelle zone
                colorer(M, i, j, val)
                val = val + 1
    return val - 2
````

*Les tests de la feuille d'exercices :*

````python
M = [[1, 1, 0, 0, 1],
     [0, 1, 0, 1, 1],
     [0, 1, 0, 0, 0],
     [1, 0, 0, 1, 1]]
assert nombre_zones(M) == 4
afficher(M)
assert nombre_zones([[0, 0], [0, 0]]) == 0
assert nombre_zones([[1, 1], [1, 1]]) == 1
assert nombre_zones([[1, 0, 1], [0, 1, 0], [1, 0, 1]]) == 5    # les diagonales ne comptent pas
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- On parcourt toute l'image avec deux boucles. Chaque fois qu'on trouve un pixel encore noir (`1`), c'est qu'il appartient à une **nouvelle** zone (sinon il aurait déjà été coloré) : on colore toute cette zone d'un coup, et on compte une zone de plus.
- On mélange ici **boucles** (pour parcourir l'image) et **récursivité** (pour remplir une zone) : chaque outil là où il est le plus simple.
- À la fin, `M` a été modifiée : chaque zone a sa propre couleur, comme dans le sujet du bac.

## Exercice 6 — Somme des chiffres  ⭐

### Question 1

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `somme_chiffres(n)` qui renvoie la somme des chiffres de l'entier positif `n`.
>
> Exemple : `somme_chiffres(2026)` renvoie `2 + 0 + 2 + 6 = 10`.
>
> *Indice : `n % 10` donne le dernier chiffre de `n` (`2026 % 10` vaut `6`), et `n // 10` donne `n` sans son dernier chiffre (`2026 // 10` vaut `202`).*

#### ✅ RÉSOLUTION

````python
def somme_chiffres(n):
    if n < 10:                   # un seul chiffre
        return n
    return n % 10 + somme_chiffres(n // 10)
````

*Les tests de la feuille d'exercices :*

````python
assert somme_chiffres(2026) == 10
assert somme_chiffres(7) == 7
assert somme_chiffres(0) == 0
assert somme_chiffres(99999) == 45
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Cas récursif :** la somme des chiffres de `2026`, c'est le dernier chiffre `6` plus la somme des chiffres de `202`.
- **Cas de base :** un nombre d'un seul chiffre (`n < 10`) : la somme de ses chiffres, c'est lui-même.
- **Terminaison :** `n // 10` a un chiffre de moins que `n`, donc on finit forcément par arriver à un seul chiffre.

## Exercice 7 — Multiplier avec des additions  ⭐⭐

📚 **Tombé au bac :** épreuve pratique 2024, sujet 18 (exercice 18.1). Le sujet demande une multiplication de deux entiers relatifs avec uniquement des additions et des soustractions ; ici on demande en plus qu'elle soit récursive.

### Question 1

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `multiplication(n1, n2)` qui renvoie le produit de deux entiers relatifs (positifs ou négatifs) `n1` et `n2`, en n'utilisant **que des additions et des soustractions** (pas de `*`).
>
> Exemples : `multiplication(3, 5)` renvoie `15`, `multiplication(-4, -8)` renvoie `32`, `multiplication(-2, 6)` renvoie `-12`, `multiplication(-2, 0)` renvoie `0`.
>
> *Indice : $n_1 \times n_2 = n_1 + n_1 \times (n_2 - 1)$. Et si `n2` est négatif, $n_1 \times n_2 = -(n_1 \times (-n_2))$.*

#### ✅ RÉSOLUTION

````python
def multiplication(n1, n2):
    if n2 == 0:
        return 0
    if n2 < 0:
        return -multiplication(n1, -n2)       # on se ramène à n2 positif
    return n1 + multiplication(n1, n2 - 1)
````

*Les tests de la feuille d'exercices :*

````python
assert multiplication(3, 5) == 15
assert multiplication(-4, -8) == 32
assert multiplication(-2, 6) == -12
assert multiplication(-2, 0) == 0
assert multiplication(7, -3) == -21
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Cas de base :** multiplier par `0` donne `0`.
- **Cas récursif (n2 positif) :** $3 \times 5 = 3 + 3 \times 4$ : on ajoute `n1`, et on diminue `n2` de 1.
- **n2 négatif :** si on diminuait `n2`, il s'éloignerait de `0` et la fonction ne s'arrêterait jamais (`RecursionError`). On se ramène donc d'abord au cas positif : $(-2) \times (-3) = -((-2) \times 3)$. Il n'y a pas besoin de traiter `n1` négatif : additionner un nombre négatif marche très bien.
- C'est un bon exemple de la question 3 de la méthode : **vérifier qu'on se rapproche du cas de base**, dans tous les cas.

## Exercice 8 — Chaînes de caractères : inverser, palindrome  ⭐⭐

### Question 1

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `inverse(ch, i)` qui renvoie la chaîne formée des caractères de `ch` à partir de l'indice `i`, **à l'envers**.
>
> Exemple : `inverse("python", 0)` renvoie `"nohtyp"`.
>
> *Rappel : `+` colle deux chaînes : `"py" + "thon"` donne `"python"`.*

#### ✅ RÉSOLUTION

````python
def inverse(ch, i):
    if i == len(ch):
        return ""
    return inverse(ch, i + 1) + ch[i]
````

*Les tests de la feuille d'exercices :*

````python
assert inverse("python", 0) == "nohtyp"
assert inverse("a", 0) == "a"
assert inverse("", 0) == ""
assert inverse("NSI", 0) == "ISN"
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Cas de base :** plus aucun caractère : la chaîne vide `""`.
- **Cas récursif :** l'inverse de `"python"`, c'est l'inverse de `"ython"` (`"nohty"`) **suivi** de `"p"`. Le caractère `ch[i]` se met donc **à la fin**.
- **Erreur fréquente :** écrire `ch[i] + inverse(ch, i + 1)` : on remet le caractère devant, et on recopie la chaîne dans le même ordre.

### Question 2

#### 📝 ÉNONCÉ

> Un **palindrome** est un mot qui se lit pareil dans les deux sens : `"radar"`, `"kayak"`, `"ressasser"`.
>
> Écrire une fonction récursive `est_palindrome(ch, g, d)` qui renvoie `True` si la partie de `ch` comprise entre les indices `g` (gauche) et `d` (droite) est un palindrome. On l'appelle avec `est_palindrome(ch, 0, len(ch) - 1)`.

#### ✅ RÉSOLUTION

````python
def est_palindrome(ch, g, d):
    if g >= d:                       # 0 ou 1 caractère : toujours un palindrome
        return True
    if ch[g] != ch[d]:
        return False
    return est_palindrome(ch, g + 1, d - 1)
````

*Les tests de la feuille d'exercices :*

````python
assert est_palindrome("radar", 0, 4)
assert est_palindrome("ressasser", 0, 8)
assert not est_palindrome("python", 0, 5)
assert est_palindrome("abba", 0, 3)
assert not est_palindrome("abca", 0, 3)
assert est_palindrome("a", 0, 0)
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Idée :** on compare le premier et le dernier caractère. S'ils sont différents, ce n'est pas un palindrome. S'ils sont égaux, on recommence avec la partie du milieu (`g + 1` et `d - 1`).
- **Cas de base :** quand les deux indices se croisent ou se rejoignent (`g >= d`), il reste 0 ou 1 caractère : c'est forcément un palindrome.
- **Terminaison :** l'écart `d - g` diminue de 2 à chaque appel.

## Exercice 9 — Conversions décimal ↔ binaire  ⭐⭐

📚 **Tombé au bac :** épreuve pratique 2024, sujet 9 (exercice 9.2) : deux fonctions récursives `dec_to_bin` et `bin_to_dec`, sans utiliser `bin` ni `int`. Le sujet utilisait le slicing pour `bin_to_dec` ; ici on utilise un indice.

### Question 1

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `dec_to_bin(n)` qui renvoie l'écriture binaire de l'entier positif `n`, sous forme de chaîne de caractères. Exemple : `dec_to_bin(25)` renvoie `"11001"`.
>
> *Indice : le dernier chiffre binaire de `n` est le reste `n % 2`, et les chiffres d'avant sont l'écriture binaire du quotient `n // 2`. Exemple : $25 = 2 \times 12 + 1$, donc l'écriture de 25 est celle de 12 (`"1100"`) suivie de `"1"`. La fonction `str` transforme un nombre en chaîne : `str(1)` vaut `"1"`.*

#### ✅ RÉSOLUTION

````python
def dec_to_bin(n):
    q = n // 2
    r = n % 2
    if q == 0:                       # n vaut 0 ou 1 : un seul chiffre
        return str(r)
    return dec_to_bin(q) + str(r)
````

*Les tests de la feuille d'exercices :*

````python
assert dec_to_bin(25) == "11001"
assert dec_to_bin(0) == "0"
assert dec_to_bin(1) == "1"
assert dec_to_bin(42) == "101010"
assert dec_to_bin(255) == "11111111"
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- C'est la méthode des **divisions successives par 2** vue en Première, écrite de façon récursive : chaque division donne un chiffre, en partant de la droite.
- Comme le reste `r` est le chiffre **de droite**, on le colle **après** l'écriture du quotient : `dec_to_bin(q) + str(r)`.
- **Cas de base :** quand le quotient vaut `0`, `n` vaut `0` ou `1` : un seul chiffre, `str(r)`.

### Question 2

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `bin_to_dec(mot, i)` qui renvoie la valeur décimale des chiffres binaires `mot[0]`, `mot[1]`, ..., `mot[i]` (de l'indice 0 jusqu'à l'indice `i` inclus). On l'appelle avec `bin_to_dec(mot, len(mot) - 1)` pour tout le mot.
>
> Exemple : `bin_to_dec("101010", 5)` renvoie `42`.
>
> *Indice : ajouter un chiffre binaire à droite **double** la valeur, puis ajoute ce chiffre : `"10101"` vaut 21, donc `"101010"` vaut $2 \times 21 + 0 = 42$.*

#### ✅ RÉSOLUTION

````python
def bin_to_dec(mot, i):
    if mot[i] == "1":                # le chiffre d'indice i, sous forme de nombre
        bit = 1
    else:
        bit = 0
    if i == 0:                       # un seul chiffre
        return bit
    return 2 * bin_to_dec(mot, i - 1) + bit
````

*Les tests de la feuille d'exercices :*

````python
assert bin_to_dec("101010", 5) == 42
assert bin_to_dec("0", 0) == 0
assert bin_to_dec("1", 0) == 1
assert bin_to_dec("11001", 4) == 25
for n in range(300):
    mot = dec_to_bin(n)
    assert bin_to_dec(mot, len(mot) - 1) == n      # les deux fonctions sont inverses
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Ici on avance **de droite à gauche** : `i` part du dernier indice et **diminue**. Cas de base : `i == 0`, il ne reste que le premier chiffre.
- **Cas récursif :** la valeur de `mot[0..i]`, c'est le double de la valeur de `mot[0..i-1]`, plus le chiffre `mot[i]`.
- On vérifie : `"101010"` → $2 \times$ valeur(`"10101"`) $+ 0$ ; valeur(`"10101"`) $= 2 \times$ valeur(`"1010"`) $+ 1$... jusqu'à valeur(`"1"`) $= 1$.

## Exercice 10 — Des mots construits avec des règles  ⭐⭐⭐

📚 **Tombé au bac :** épreuve écrite 2022, Polynésie, sujet 1, exercice 1 (« programmation, et principalement la récursivité »). Questions reprises du sujet ; la fonction `raccourcir` est donnée ici écrite avec une boucle.

On s'intéresse à des chaînes de caractères construites selon des règles :
- **Règle A :** une chaîne suit la règle A si elle est égale à `"a"`, **ou** si elle est de la forme `"a" + chaine + "a"` où `chaine` suit la règle A. *(Exemples : `"a"`, `"aaa"`, `"aaaaa"`...)*
- **Règle B :** une chaîne suit la règle B si elle est de la forme `"b" + chaine + "b"` où `chaine` suit la règle A **ou** la règle B. *(Exemples : `"bab"`, `"baaab"`, `"bbabb"`...)*

La fonction `choice` du module `random` choisit un élément au hasard dans une liste : `choice([True, False])` renvoie `True` ou `False` au hasard.

La fonction `A()` ci-dessous renvoie une chaîne construite suivant la règle A, en choisissant au hasard entre les deux cas de la règle :
```python
def A():
    if choice([True, False]):
        return "a"
    else:
        return "a" + A() + "a"
```

### Question 1 ✍️

#### 📝 ÉNONCÉ

> (a) Cette fonction est-elle récursive ? Justifier.

> (b) La fonction `choice([True, False])` peut renvoyer `False` un très grand nombre de fois de suite. Expliquer pourquoi ce cas de figure amènerait à une erreur d'exécution.

#### ✅ RÉSOLUTION

**(a)** Oui : la fonction `A` s'appelle elle-même (dans le `else`), donc elle est récursive.

**(b)** Chaque `False` provoque un nouvel appel récursif **en attente** (la concaténation `"a" + A() + "a"` ne peut se faire que quand `A()` a renvoyé sa valeur). Les appels s'empilent dans la pile d'appels, dont la taille est limitée (environ 1000 en Python). Au-delà, Python s'arrête avec l'erreur `RecursionError: maximum recursion depth exceeded`.

#### 💡 EXPLICATION PÉDAGOGIQUE

- Une réponse « récursive » doit toujours être **justifiée** : « elle s'appelle elle-même ».
- Pour (b), le mot-clé attendu est la **pile d'appels** et sa **taille limitée** (partie 4 du cours). On peut citer l'erreur `RecursionError`.

### Question 2

#### 📝 ÉNONCÉ

> On considère une deuxième version de la fonction `A`, qui prend en paramètre un entier `n`. Si `n` est négatif ou nul, la fonction renvoie `"a"`. Sinon, elle choisit au hasard entre les deux cas de la règle A, en appelant récursivement `A` avec `n - 1`.

**(b)** À chaque appel, soit la fonction s'arrête (si `n <= 0` ou si `choice` renvoie `True`), soit elle s'appelle avec `n - 1`. Pour `n < 50`, il y a donc au plus 50 appels, bien moins que la limite de la pile.

### Question 3 ✍️

#### 📝 ÉNONCÉ

> On donne la fonction récursive `B(n)`, qui renvoie une chaîne construite suivant la règle B :
> ```python
> def B(n):
>     if n <= 0 or choice([True, False]):
>         return "b" + A(n - 1) + "b"
>     else:
>         return "b" + B(n - 1) + "b"
> ```
> On admet que les appels `A(-1)` et `A(0)` renvoient `"a"`, que `A(1)` renvoie `"a"` ou `"aaa"`, et que `A(2)` renvoie `"a"`, `"aaa"` ou `"aaaaa"`.
>
> Donner toutes les chaînes possibles renvoyées par les appels `B(0)`, `B(1)` et `B(2)`.

#### ✅ RÉSOLUTION

- `B(0)` : `n <= 0`, donc on renvoie `"b" + A(-1) + "b"`, c'est-à-dire **`"bab"`**.
- `B(1)` : soit `"b" + A(0) + "b"` = `"bab"`, soit `"b" + B(0) + "b"` = `"bbabb"`. Donc **`"bab"` ou `"bbabb"`**.
- `B(2)` : soit `"b" + A(1) + "b"`, qui donne `"bab"` ou `"baaab"` ; soit `"b" + B(1) + "b"`, qui donne `"bbabb"` ou `"bbbabbb"`. Donc **`"bab"`, `"baaab"`, `"bbabb"` ou `"bbbabbb"`**.

````python
def B(n):
    if n <= 0 or choice([True, False]):
        return "b" + A(n - 1) + "b"
    else:
        return "b" + B(n - 1) + "b"


resultats = []
for k in range(2000):
    mot = B(2)
    if mot not in resultats:
        resultats.append(mot)
print(resultats)
````

#### 📝 ÉNONCÉ

> On dispose d'une fonction `raccourcir` qui prend une chaîne de longueur au moins 2 et renvoie la chaîne sans son premier ni son dernier caractère (elle est donnée ci-dessous). Par exemple, `raccourcir("abricot")` renvoie `"brico"` et `raccourcir("ab")` renvoie `""`.
>
> (a) Compléter le code de la fonction `regleA` pour qu'elle renvoie `True` si la chaîne suit la règle A, et `False` sinon.
> ```python
> def regleA(chaine):
>     n = len(chaine)
>     if n >= 2:
>         return chaine[0] == "a" and chaine[n - 1] == "a" and regleA(...)
>     else:
>         return chaine == ...
> ```
> (b) Écrire une fonction `regleB(chaine)` qui renvoie `True` si la chaîne suit la règle B, et `False` sinon.

#### ✅ RÉSOLUTION

````python
def raccourcir(chaine):
    res = ""
    for k in range(1, len(chaine) - 1):
        res = res + chaine[k]
    return res


def regleA(chaine):
    n = len(chaine)
    if n >= 2:
        return chaine[0] == "a" and chaine[n - 1] == "a" and regleA(raccourcir(chaine))
    else:
        return chaine == "a"


def regleB(chaine):
    n = len(chaine)
    if n >= 2:
        milieu = raccourcir(chaine)
        return chaine[0] == "b" and chaine[n - 1] == "b" and (regleA(milieu) or regleB(milieu))
    else:
        return False
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **On traduit la règle mot pour mot :** « de la forme `"a" + chaine + "a"` avec `chaine` qui suit la règle A » devient : premier caractère `"a"`, dernier caractère `"a"`, et le milieu (`raccourcir(chaine)`) suit la règle A.
- **Cas de base de `regleA` :** une chaîne de moins de 2 caractères suit la règle A seulement si c'est `"a"`.
- **Cas de base de `regleB` :** une chaîne de moins de 2 caractères ne peut jamais suivre la règle B (elle commence et finit par `"b"` autour d'une autre chaîne non vide).
- `regleB` appelle `regleA` **et** `regleB` : la règle B fait référence aux deux règles.
- **Terminaison :** `raccourcir` enlève 2 caractères, donc la longueur diminue à chaque appel.

## Exercice 11 — Rendu de monnaie  ⭐⭐⭐

📚 **Tombé au bac :** épreuve pratique 2024, sujet 36 (exercice 36.2) : `rendu_glouton(a_rendre, rang)`, avec par exemple `rendu_glouton(67, 0)` qui renvoie `[50, 10, 5, 2]`.

On dispose de pièces et de billets dont les valeurs sont rangées dans la liste `valeurs` (par ordre décroissant). Pour rendre une somme, on utilise l'algorithme **glouton** : on rend toujours la **plus grande** valeur possible, tant qu'elle ne dépasse pas ce qu'il reste à rendre.

```python
valeurs = [100, 50, 20, 10, 5, 2, 1]
```

*Rappel : `+` colle aussi deux listes : `[50] + [10, 5]` donne `[50, 10, 5]`.*

### Question 1

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `rendu_glouton(a_rendre, rang)` qui renvoie la liste des pièces et billets à rendre pour la somme `a_rendre`, en n'utilisant que les valeurs `valeurs[rang]`, `valeurs[rang + 1]`, ... On l'appelle avec `rang = 0`.
>
> *Indice : si `valeurs[rang]` ne dépasse pas `a_rendre`, on la rend, et il reste `a_rendre - valeurs[rang]` à rendre (avec la même valeur possible encore une fois). Sinon, on passe à la valeur suivante (`rang + 1`).*

#### ✅ RÉSOLUTION

````python
valeurs = [100, 50, 20, 10, 5, 2, 1]


def rendu_glouton(a_rendre, rang):
    if a_rendre == 0:
        return []
    v = valeurs[rang]
    if v <= a_rendre:
        return [v] + rendu_glouton(a_rendre - v, rang)      # on rend v, on garde le même rang
    return rendu_glouton(a_rendre, rang + 1)                # v est trop grande : on passe à la suivante
````

*Les tests de la feuille d'exercices :*

````python
assert rendu_glouton(67, 0) == [50, 10, 5, 2]
assert rendu_glouton(291, 0) == [100, 100, 50, 20, 20, 1]
assert rendu_glouton(0, 0) == []
assert rendu_glouton(8, 0) == [5, 2, 1]
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Cas de base :** il n'y a plus rien à rendre : la liste vide `[]`.
- **Deux cas récursifs :**
  - la valeur `v` convient : on la rend (`[v] + ...`) et on rend le reste, **sans changer de rang** (on peut avoir besoin de plusieurs fois la même valeur, comme deux billets de 100) ;
  - elle est trop grande : on ne rend rien et on passe à la valeur suivante.
- **Terminaison :** à chaque appel, soit `a_rendre` diminue, soit `rang` augmente. Comme la dernière valeur est `1`, on finit toujours par tout rendre avant de dépasser la fin de la liste.

## Exercice 12 — Les chiffres romains  ⭐⭐⭐

📚 **Tombé au bac :** épreuve pratique 2024, sujet 30 (exercice 30.2) : `traduire_romain(nombre)`. Le sujet utilisait le slicing ; ici on utilise un indice.

En chiffres romains, `I = 1`, `V = 5`, `X = 10`, `L = 50`, `C = 100`, `D = 500`, `M = 1000`. On les stocke dans un dictionnaire :
```python
romains = {"I": 1, "V": 5, "X": 10, "L": 50, "C": 100, "D": 500, "M": 1000}
```
On lit un nombre de gauche à droite. Un symbole **ajoute** sa valeur, **sauf** s'il est suivi d'un symbole de valeur plus grande : dans ce cas, il la **retranche**. Exemples : `"XVI"` vaut $10 + 5 + 1 = 16$, mais `"XIV"` vaut $10 + (5 - 1) = 14$, et `"CXLII"` vaut $100 + (50 - 10) + 1 + 1 = 142$.

### Question 1

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `traduire_romain(nombre, i)` qui renvoie la valeur de la partie de la chaîne `nombre` qui commence à l'indice `i`. On l'appelle avec `traduire_romain(nombre, 0)`.
>
> *Indice : cas de base quand `i` est l'indice du dernier symbole. Sinon, comparer la valeur du symbole `nombre[i]` à celle du suivant `nombre[i + 1]`.*

#### ✅ RÉSOLUTION

````python
romains = {"I": 1, "V": 5, "X": 10, "L": 50, "C": 100, "D": 500, "M": 1000}


def traduire_romain(nombre, i):
    valeur = romains[nombre[i]]
    if i == len(nombre) - 1:                         # dernier symbole
        return valeur
    if valeur >= romains[nombre[i + 1]]:
        return valeur + traduire_romain(nombre, i + 1)
    return traduire_romain(nombre, i + 1) - valeur
````

*Les tests de la feuille d'exercices :*

````python
assert traduire_romain("XIV", 0) == 14
assert traduire_romain("CXLII", 0) == 142
assert traduire_romain("MMXXIV", 0) == 2024
assert traduire_romain("MCMXCIX", 0) == 1999
assert traduire_romain("I", 0) == 1
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Cas de base :** le dernier symbole compte toujours positivement (il n'est suivi de rien).
- **Cas récursif :** on traduit la suite de la chaîne (appel avec `i + 1`), puis on **ajoute** ou on **retranche** la valeur du symbole courant selon le symbole qui le suit.
- Exemple : `"XIV"` → `10 + traduire("IV")` ; `"IV"` → `traduire("V") - 1 = 4` ; total `14`.

## Exercice 13 — La suite de Fibonacci  ⭐⭐⭐

📚 **Tombé au bac :** épreuve pratique 2024, sujet 28 (exercice 28.1) : `fibonacci(n)`, avec par exemple `fibonacci(25) = 75025`.

La suite de Fibonacci commence par `1, 1`, puis chaque terme est la **somme des deux précédents** : `1, 1, 2, 3, 5, 8, 13, 21, ...`

On note `fibonacci(n)` le terme d'indice `n`, avec `fibonacci(1) = 1` et `fibonacci(2) = 1`.

### Question 1

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `fibonacci(n)` (pour `n >= 1`).

#### ✅ RÉSOLUTION

````python
def fibonacci(n):
    if n == 1 or n == 2:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)
````

*Les tests de la feuille d'exercices :*

````python
assert fibonacci(1) == 1
assert fibonacci(2) == 1
assert fibonacci(6) == 8
assert fibonacci(10) == 55
assert fibonacci(25) == 75025
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- La définition est elle-même récursive : il suffit de la recopier.
- Il faut **deux** cas de base, car la fonction a besoin des **deux** termes précédents : avec seulement `n == 1`, l'appel `fibonacci(2)` appellerait `fibonacci(0)`, puis `fibonacci(-1)`... sans jamais s'arrêter.

### Question 2 ✍️

#### 📝 ÉNONCÉ

> On compte le nombre d'appels avec une variable globale `nb_appels`. Exécuter la cellule ci-dessous, puis expliquer pourquoi le nombre d'appels augmente aussi vite. Essayer `fibonacci_compte(35)` : que se passe-t-il ?

#### ✅ RÉSOLUTION

**Explication :** chaque appel en déclenche **deux** autres, et les mêmes calculs sont refaits un très grand nombre de fois. Par exemple `fibonacci(5)` calcule `fibonacci(4)` et `fibonacci(3)`, mais `fibonacci(4)` recalcule **lui aussi** `fibonacci(3)` :
```
                    fib(5)
                 /          \
            fib(4)          fib(3)
           /      \         /     \
       fib(3)   fib(2)   fib(2)  fib(1)
       /    \
   fib(2)  fib(1)
```
Le nombre d'appels est multiplié par environ 1,6 à chaque fois que `n` augmente de 1 : pour `n = 35`, il faut presque 30 millions d'appels, et le calcul prend plusieurs secondes. Pour `n = 50`, il faudrait des heures.

````python
nb_appels = 0

def fibonacci_compte(n):
    global nb_appels
    nb_appels = nb_appels + 1
    if n == 1 or n == 2:
        return 1
    return fibonacci_compte(n - 1) + fibonacci_compte(n - 2)


for n in [5, 10, 15, 20, 25, 30]:
    nb_appels = 0
    resultat = fibonacci_compte(n)
    print("fibonacci(" + str(n) + ") =", resultat, "  appels :", nb_appels)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- C'est **le** piège classique de la récursivité : deux appels récursifs **sur des sous-problèmes qui se chevauchent** refont sans cesse le même travail.
- Comparaison avec l'exponentiation rapide : là aussi, la « version maladroite » `power_dac(x, n // 2) * power_dac(x, n // 2)` calculait deux fois la même chose ; il suffisait de stocker le résultat dans une variable. Ici, les deux appels sont **différents** (`n - 1` et `n - 2`), donc ce n'est pas aussi simple.
- **Solutions :** une simple **boucle** qui garde les deux derniers termes (question 3), ou la **mémoïsation** (retenir dans un dictionnaire les résultats déjà calculés), qu'on voit en Terminale avec la programmation dynamique.

### Question 3

#### 📝 ÉNONCÉ

> Écrire une fonction **non récursive** `fibonacci_boucle(n)`, avec une boucle, qui calcule le même résultat très rapidement, même pour `n = 100`.
>
> *Indice : garder dans deux variables les deux derniers termes calculés.*

#### ✅ RÉSOLUTION

````python
def fibonacci_boucle(n):
    avant_dernier = 1
    dernier = 1
    for k in range(n - 2):
        suivant = avant_dernier + dernier
        avant_dernier = dernier
        dernier = suivant
    return dernier


print(fibonacci_boucle(100))
````

*Les tests de la feuille d'exercices :*

````python
for n in range(1, 21):
    assert fibonacci_boucle(n) == fibonacci(n)
assert fibonacci_boucle(100) == 354224848179261915075
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- La boucle calcule chaque terme **une seule fois**, en partant du début : `n - 2` additions seulement.
- La récursivité n'est pas toujours la meilleure solution : elle est élégante, mais il faut **réfléchir au nombre d'appels**.

## Exercice 14 — Chemins dans une grille  ⭐⭐⭐

📚 **Tombé au bac :** épreuve écrite 2024, Centres étrangers 1, jour 2, exercice 1 (partie C, questions 7 à 9).

On se déplace dans une grille rectangulaire de `m` colonnes et `n` lignes, de la case **en haut à gauche** (S) à la case **en bas à droite** (E). Les seuls déplacements autorisés sont **une case vers la droite** (D) ou **une case vers le bas** (B).

```
S → → ↓          un chemin possible dans une grille 4 × 3 : D D B D B
    · ↓ → ↓
    · · · E
```

On note `N(m, n)` le nombre de chemins différents dans une grille de `m` colonnes et `n` lignes.

### Question 1 ✍️

#### 📝 ÉNONCÉ

> Pour une grille d'une seule ligne (dimension `m × 1`), justifier qu'il y a un seul chemin : `N(m, 1) = 1`. De même, `N(1, n) = 1`.

#### ✅ RÉSOLUTION

Dans une grille d'une seule ligne, on ne peut jamais descendre : le seul chemin consiste à aller tout droit vers la droite jusqu'à E. Il y a donc un seul chemin. De même, dans une grille d'une seule colonne, on ne peut que descendre.

#### 💡 EXPLICATION PÉDAGOGIQUE

- Ce sont les **cas de base** de la future fonction récursive.

### Question 2 ✍️

#### 📝 ÉNONCÉ

> Justifier que `N(m, n) = N(m - 1, n) + N(m, n - 1)`.

#### ✅ RÉSOLUTION

Depuis la case de départ, le premier déplacement est soit **vers la droite**, soit **vers le bas** :
- si on va à droite, il reste à aller de cette nouvelle case jusqu'à E : c'est un chemin dans une grille d'**une colonne de moins**, il y en a `N(m - 1, n)` ;
- si on descend, il reste un chemin dans une grille d'**une ligne de moins** : il y en a `N(m, n - 1)`.

Ces deux familles de chemins sont différentes (le premier pas n'est pas le même) et elles contiennent tous les chemins. Donc `N(m, n) = N(m - 1, n) + N(m, n - 1)`.

#### 💡 EXPLICATION PÉDAGOGIQUE

- Raisonnement typique : on **sépare en cas** selon le **premier** choix, et chaque cas est **le même problème en plus petit**.
- On peut aussi raisonner sur le **dernier** déplacement (on arrive en E par la gauche ou par le haut) : on obtient la même formule.

### Question 3

#### 📝 ÉNONCÉ

> En utilisant les questions précédentes, écrire une fonction récursive `nombre_chemins(m, n)` qui renvoie le nombre de chemins possibles dans une grille de dimension `m × n`.

#### ✅ RÉSOLUTION

````python
def nombre_chemins(m, n):
    if m == 1 or n == 1:
        return 1
    return nombre_chemins(m - 1, n) + nombre_chemins(m, n - 1)
````

*Les tests de la feuille d'exercices :*

````python
assert nombre_chemins(1, 5) == 1
assert nombre_chemins(5, 1) == 1
assert nombre_chemins(2, 2) == 2
assert nombre_chemins(3, 3) == 6
assert nombre_chemins(4, 3) == 10
assert nombre_chemins(3, 4) == 10
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Une traduction directe : cas de base (question 1), puis formule (question 2).
- **Terminaison :** à chaque appel, `m` ou `n` diminue de 1, et on s'arrête dès que l'un des deux vaut 1.
- Comme pour Fibonacci, il y a **deux** appels récursifs, et les mêmes grilles sont recalculées plusieurs fois : `nombre_chemins(15, 15)` fait déjà des dizaines de millions d'appels. (Pour aller plus loin : le résultat est le coefficient binomial $\binom{m+n-2}{m-1}$, qu'on retrouve dans le triangle de Pascal.)

## Exercice 15 — Recherche dichotomique récursive  ⭐⭐⭐

📚 **Tombé au bac :** épreuve pratique 2024, sujet 18 (exercice 18.2) : `chercher(tab, x, i, j)`.

On cherche une valeur `x` dans une liste `tab` **triée** dans l'ordre croissant. La **recherche dichotomique** regarde l'élément du milieu :
- s'il est égal à `x`, on a trouvé ;
- s'il est plus petit que `x`, on cherche dans la moitié droite ;
- s'il est plus grand, on cherche dans la moitié gauche.

### Question 1

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `chercher(tab, x, i, j)` qui renvoie un indice de `x` dans la partie de `tab` comprise entre les indices `i` et `j` (inclus), ou `None` si `x` n'y est pas. On l'appelle avec `chercher(tab, x, 0, len(tab) - 1)`.
>
> Exemples avec `tab = [1, 5, 6, 6, 9, 12]` : `chercher(tab, 9, 0, 5)` renvoie `4`, `chercher(tab, 6, 0, 5)` renvoie `2`, `chercher(tab, 7, 0, 5)` renvoie `None`.

#### ✅ RÉSOLUTION

````python
def chercher(tab, x, i, j):
    if i > j:                        # zone de recherche vide : x n'y est pas
        return None
    m = (i + j) // 2
    if tab[m] < x:
        return chercher(tab, x, m + 1, j)       # moitié droite
    if tab[m] > x:
        return chercher(tab, x, i, m - 1)       # moitié gauche
    return m
````

*Les tests de la feuille d'exercices :*

````python
tab = [1, 5, 6, 6, 9, 12]
assert chercher(tab, 9, 0, 5) == 4
assert chercher(tab, 6, 0, 5) == 2
assert chercher(tab, 7, 0, 5) == None
assert chercher(tab, 1, 0, 5) == 0 and chercher(tab, 12, 0, 5) == 5
assert chercher(tab, 0, 0, 5) == None and chercher(tab, 20, 0, 5) == None
assert chercher([], 3, 0, -1) == None
grand = []
for k in range(0, 2000000, 2):
    grand.append(k)
assert chercher(grand, 1234568, 0, len(grand) - 1) == 617284
assert chercher(grand, 1234567, 0, len(grand) - 1) == None
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Deux cas de base :** la zone est vide (`i > j`, on n'a pas trouvé), ou l'élément du milieu est `x` (on a trouvé).
- **Cas récursif :** on relance la recherche sur **une seule moitié**, en déplaçant `i` ou `j`. On exclut l'indice `m`, déjà testé (`m + 1` et `m - 1`) : sinon la zone pourrait ne plus diminuer et la fonction tournerait à l'infini.
- **Coût :** la zone est divisée par 2 à chaque appel. Pour une liste d'un million d'éléments, il faut au plus une vingtaine d'appels (car $2^{20} \approx 1\,000\,000$), contre un million pour un parcours un par un.

## Exercice 16 — Trouver l'intrus  ⭐⭐⭐⭐

📚 **Tombé au bac :** épreuve pratique 2024, sujet 40 (exercice 40.2) : `trouver_intrus(tab, g, d)`.

Dans une liste, chaque valeur apparaît **trois fois de suite**, sauf **une seule** valeur, l'intrus, qui n'apparaît qu'**une fois**. Exemple :
```python
tab_a = [3, 3, 3, 9, 9, 9, 1, 1, 1, 7, 2, 2, 2, 4, 4, 4, 8, 8, 8, 5, 5, 5]      # l'intrus est 7
```
On veut trouver l'intrus **sans parcourir toute la liste**, par dichotomie. On découpe la liste en « paquets » de 3 à partir du début : `[3, 3, 3]`, `[9, 9, 9]`, ... Avant l'intrus, chaque paquet contient trois valeurs égales. À partir de l'intrus, les paquets sont **décalés** : le premier élément d'un paquet n'est plus égal au deuxième.

### Question 1

#### 📝 ÉNONCÉ

> Écrire une fonction récursive `trouver_intrus(tab, g, d)` qui renvoie l'intrus situé entre les indices `g` et `d` (inclus), où `g` est le début d'un paquet. On l'appelle avec `trouver_intrus(tab, 0, len(tab) - 1)`.
>
> *Indice : s'il ne reste qu'un élément (`g == d`), c'est l'intrus. Sinon, on calcule le nombre de paquets complets `(d - g) // 3`, on prend le paquet du milieu qui commence à l'indice `indice = g + 3 * (nombre_de_paquets // 2)`, et on regarde si `tab[indice]` est égal à `tab[indice + 1]`.*

#### ✅ RÉSOLUTION

````python
def trouver_intrus(tab, g, d):
    if g == d:
        return tab[g]
    nombre_de_paquets = (d - g) // 3
    indice = g + 3 * (nombre_de_paquets // 2)
    if tab[indice] != tab[indice + 1]:
        return trouver_intrus(tab, g, indice)        # le décalage a déjà eu lieu : l'intrus est à gauche (ou en indice)
    return trouver_intrus(tab, indice + 3, d)        # paquet normal : l'intrus est après ce paquet
````

*Les tests de la feuille d'exercices :*

````python
tab_a = [3, 3, 3, 9, 9, 9, 1, 1, 1, 7, 2, 2, 2, 4, 4, 4, 8, 8, 8, 5, 5, 5]
tab_b = [8, 5, 5, 5, 9, 9, 9, 18, 18, 18, 3, 3, 3]
tab_c = [5, 5, 5, 1, 1, 1, 0, 0, 0, 6, 6, 6, 3, 8, 8, 8]
assert trouver_intrus(tab_a, 0, len(tab_a) - 1) == 7
assert trouver_intrus(tab_b, 0, len(tab_b) - 1) == 8
assert trouver_intrus(tab_c, 0, len(tab_c) - 1) == 3
assert trouver_intrus([4], 0, 0) == 4
# Vérification sur toutes les positions possibles de l'intrus
for nb in range(0, 12):
    for position in range(nb + 1):
        t = []
        for k in range(nb + 1):
            if k == position:
                t.append(-1)                 # l'intrus
            if k < nb:
                for r in range(3):
                    t.append(k)
        assert trouver_intrus(t, 0, len(t) - 1) == -1
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **L'observation clé :** si le paquet qui commence en `indice` est « normal » (`tab[indice] == tab[indice + 1]`), l'intrus est **après** ce paquet : on cherche à partir de `indice + 3`. Sinon, le décalage a déjà commencé : l'intrus est **à `indice` ou avant**.
- On garde toujours `g` au **début d'un paquet**, pour que les paquets restent bien alignés.
- La zone est à peu près divisée par 2 à chaque appel : c'est une dichotomie, avec un coût logarithmique, alors qu'un parcours complet regarderait tous les éléments.
- C'est un exercice difficile : le plus simple est de dérouler l'algorithme **à la main** sur `tab_a`, en notant `g`, `d` et `indice` à chaque appel.
