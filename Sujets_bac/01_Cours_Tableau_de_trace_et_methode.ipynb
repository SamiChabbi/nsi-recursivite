# 2. Le tableau de trace d'une boucle

Avant la récursivité, revoyons la grille que vous utilisez en cours. Elle sert à **suivre les variables** d'un programme, étape par étape.

**Les règles :**
1. une **colonne par variable** qui change (plus, si c'est utile, une colonne pour la **condition** du `while`) ;
2. une **première ligne** avec les valeurs **avant** la boucle ;
3. puis **une ligne par tour de boucle** : on écrit la valeur de **toutes** les variables **à la fin** du tour ;
4. on s'arrête quand la boucle s'arrête (pour un `while` : quand la condition devient fausse).

> 💡 On ne « passe » pas d'une colonne à l'autre : on avance **ligne par ligne**. Une ligne, c'est une **photo** de toutes les variables à un instant donné. Si une variable ne change pas pendant un tour, on recopie sa valeur.

### Exemple 1 : une boucle `for`
On cherche ce que renvoie `somme_entiers(3)` (sujet 0, 2023).

````python
def somme_entiers(n):
    s = 0
    for i in range(n + 1):
        s = s + i
    return s
````

| Étape | `i` | `s` |
|:---|:---:|:---:|
| avant la boucle | — | 0 |
| fin du tour 1 | 0 | 0 |
| fin du tour 2 | 1 | 1 |
| fin du tour 3 | 2 | 3 |
| fin du tour 4 | 3 | 6 |

`range(4)` donne `0, 1, 2, 3` : il y a **4 tours**. À la fin, la fonction renvoie `s`, c'est-à-dire **6**.

### Exemple 2 : une boucle `while`
On cherche ce que fait `f1(7)` (Métropole, septembre 2024).

````python
def f1(n):
    i = n
    while i != 10:
        i = i + 1
    return i
````

| Étape | `i` | condition `i != 10` |
|:---|:---:|:---:|
| avant la boucle | 7 | `True` : on entre |
| fin du tour 1 | 8 | `True` : on continue |
| fin du tour 2 | 9 | `True` : on continue |
| fin du tour 3 | 10 | `False` : **on sort** |

Les valeurs successives de `i` sont 7, 8, 9, 10. L'appel **se termine** et renvoie **10**.

Avec `f1(12)`, `i` vaut 12, 13, 14, 15, 16… et n'est **jamais** égal à 10 : le tableau ne s'arrête jamais, l'appel **ne se termine pas**.

````python
# Vérification : on affiche i à chaque tour (une "trace" automatique)
def f1_trace(n):
    i = n
    print("avant la boucle : i =", i)
    while i != 10:
        i = i + 1
        print("fin du tour : i =", i)
    return i


print("renvoie", f1_trace(7))
````

# 3. Le tableau de trace d'une fonction récursive

Dans une fonction récursive, il n'y a pas de boucle : ce sont les **appels** qui se répètent. On fait donc **une ligne par appel**.

**Les règles :**
1. **une ligne par appel**, en commençant par l'appel de départ ;
2. pour chaque appel, on écrit la valeur des **paramètres**, et si c'est le **cas de base** ou non ;
3. tant que ce n'est pas le cas de base, on écrit **ce qu'il faut calculer** (avec l'appel suivant **en attente**), et on passe à la ligne suivante : c'est la **descente** ⬇️ ;
4. quand on arrive au cas de base, on connaît enfin une valeur ;
5. on remplit alors la colonne **« renvoie » en remontant**, du bas vers le haut : chaque ligne utilise la valeur de la ligne du dessous. C'est la **remontée** ⬆️.

### Exemple : `somme_entiers_rec(3)` (sujet 0, 2023)

````python
def somme_entiers_rec(n):
    if n == 0:
        return 0
    else:
        print(n)  # pour vérification
        return n + somme_entiers_rec(n - 1)
````

| Appel | `n` | Cas de base ? | Affiche | Ce qu'il faut calculer | Renvoie (on remplit en remontant ⬆️) |
|:---|:---:|:---:|:---:|:---|:---:|
| `somme_entiers_rec(3)` ⬇️ | 3 | non | `3` | `3 + somme_entiers_rec(2)` | 3 + 3 = **6** |
| `somme_entiers_rec(2)` ⬇️ | 2 | non | `2` | `2 + somme_entiers_rec(1)` | 2 + 1 = **3** ⬆️ |
| `somme_entiers_rec(1)` ⬇️ | 1 | non | `1` | `1 + somme_entiers_rec(0)` | 1 + 0 = **1** ⬆️ |
| `somme_entiers_rec(0)` | 0 | **oui** | rien | — | **0** ⬆️ |

**Comment on l'a rempli :**
- on a d'abord rempli les 5 premières colonnes **de haut en bas** (la descente) ;
- puis la dernière colonne **de bas en haut** (la remontée) : 0, puis 1 + 0 = 1, puis 2 + 1 = 3, puis 3 + 3 = 6.

**Réponses aux questions du sujet :**
- ce qui est **affiché** : on lit la colonne « Affiche » de haut en bas : `3`, `2`, `1` ;
- la **valeur renvoyée** : la case en haut à droite, **6** ;
- le **nombre d'appels** : le nombre de lignes, **4** (en comptant l'appel de départ).

````python
res = somme_entiers_rec(3)
print("res =", res)
````

> ⚠️ **Erreur fréquente :** vouloir calculer la valeur renvoyée **pendant** la descente. C'est impossible : au moment de l'appel `somme_entiers_rec(3)`, on ne connaît pas encore `somme_entiers_rec(2)`. L'appel reste **en attente** (on le voit dans la colonne « Ce qu'il faut calculer »), et on ne peut finir le calcul qu'en **remontant**.

> 💡 C'est exactement la **pile d'appels** du cours : la descente empile les appels, la remontée les dépile.

# 4. Prévoir ce qu'affiche un `print`

Une question très fréquente : la fonction contient un `print`, et on demande **ce qui est affiché**. Tout dépend d'**où** est le `print` par rapport à l'appel récursif.

| Le `print` est… | Il s'exécute… | On lit les affichages… |
|:---|:---|:---|
| **avant** l'appel récursif | pendant la **descente** ⬇️ | de haut en bas dans le tableau |
| **après** l'appel récursif | pendant la **remontée** ⬆️ | de bas en haut dans le tableau |

Comparer ces deux fonctions, qui ne diffèrent que par l'ordre de deux lignes :

````python
def avant(n):
    if n > 0:
        print(n)       # avant l'appel
        avant(n - 1)


def apres(n):
    if n > 0:
        apres(n - 1)
        print(n)       # après l'appel


print("avant(3) :")
avant(3)
print("apres(3) :")
apres(3)
````

**Pourquoi `apres(3)` affiche-t-il `1, 2, 3` ?** L'appel `apres(3)` commence par appeler `apres(2)`, et ne fera son `print(3)` que quand `apres(2)` sera **complètement terminé**. Or `apres(2)` attend lui-même `apres(1)`, etc. Le premier `print` exécuté est donc celui de l'appel le plus profond, `apres(1)`.

> 💡 **Méthode :** faire le tableau de la partie 3 avec une colonne « Affiche avant » et une colonne « Affiche après ». On lit la première de haut en bas, **puis** la deuxième de bas en haut.

> ⚠️ Lire **attentivement** la condition d'arrêt : dans le sujet 0 2023, `compte_rebours` teste `if n >= 0`, donc il affiche aussi **0** avant de s'arrêter (au moment de l'appel `compte_rebours(-1)`).

# 5. L'arbre des appels (quand la fonction s'appelle plusieurs fois)

Quand une fonction fait **deux appels récursifs ou plus** (tri fusion, Fibonacci, tri de Stooge…), un tableau ne suffit plus : on dessine un **arbre des appels**.

**Les règles :**
1. on écrit l'appel de départ en haut ;
2. sous chaque appel, on écrit les appels qu'il fait, **de gauche à droite dans l'ordre du code** ;
3. on s'arrête sur les appels qui sont des **cas de base** (ils ne font aucun appel).

**L'ordre d'exécution :** Python **termine complètement** l'appel de gauche (avec tous ses descendants) **avant** de commencer celui de droite. Les affichages se lisent donc en « descendant toujours à gauche d'abord ».

### Exemple : le tri fusion avec un `print` (Métropole 2021)

````python
def fusion(L1, L2):
    L = []
    i1 = 0
    i2 = 0
    while i1 < len(L1) or i2 < len(L2):
        if i1 >= len(L1):
            L.append(L2[i2])
            i2 = i2 + 1
        elif i2 >= len(L2):
            L.append(L1[i1])
            i1 = i1 + 1
        elif L1[i1] <= L2[i2]:
            L.append(L1[i1])
            i1 = i1 + 1
        else:
            L.append(L2[i2])
            i2 = i2 + 1
    return L


def moitie_gauche(L):
    res = []
    for i in range(len(L) // 2):
        res.append(L[i])
    return res


def moitie_droite(L):
    res = []
    for i in range(len(L) // 2, len(L)):
        res.append(L[i])
    return res


def tri_fusion(L):
    n = len(L)
    if n <= 1:
        return L
    print(L)
    mg = moitie_gauche(L)
    md = moitie_droite(L)
    L1 = tri_fusion(mg)
    L2 = tri_fusion(md)
    return fusion(L1, L2)
````

L'arbre des appels de `tri_fusion([7, 4, 2, 1])`, avec l'**ordre** dans lequel les appels commencent :

```
                       ① tri_fusion([7, 4, 2, 1])
                      /                          \
       ② tri_fusion([7, 4])                 ⑤ tri_fusion([2, 1])
          /             \                      /             \
③ tri_fusion([7])  ④ tri_fusion([4])  ⑥ tri_fusion([2])  ⑦ tri_fusion([1])
   (cas de base)      (cas de base)      (cas de base)      (cas de base)
```

- **Nombre d'appels :** on compte les cases : **7** (en comptant l'appel de départ), donc 6 appels récursifs.
- **Affichages :** le `print(L)` est fait par les appels qui ne sont **pas** des cas de base, dans l'ordre ①, ②, ⑤ : `[7, 4, 2, 1]`, puis `[7, 4]`, puis `[2, 1]`.
- **Valeurs renvoyées :** on remonte l'arbre : ② renvoie `[4, 7]`, ⑤ renvoie `[1, 2]`, et ① renvoie la fusion des deux : `[1, 2, 4, 7]`.

````python
print("renvoie", tri_fusion([7, 4, 2, 1]))
````

> 💡 **Pour compter les appels** sans dessiner tout l'arbre, on regarde ce qui se répète : chaque appel qui n'est pas un cas de base fait le **même nombre** d'appels (2 pour le tri fusion, 3 pour le tri de Stooge). On compte niveau par niveau, et on fait la somme.

> ⚠️ Toujours vérifier dans l'énoncé si on doit **compter l'appel initial** ou non : « sans compter l'appel initial » (Amérique du Nord 2024) ≠ « on veillera à compter l'appel initial » (Liban 2023).

# 6. Compléter ou écrire une fonction récursive

C'est la question la plus fréquente. On applique la **méthode en 3 questions** du cours :

1. **Cas de base :** pour quelle(s) valeur(s) la réponse est immédiate ? Que renvoie-t-on ?
2. **Plus petit problème :** sur quoi fait-on l'appel récursif (`n - 1` ? une moitié du tableau ? l'indice suivant ?) ;
3. **Combiner :** comment obtient-on la réponse à partir du résultat de l'appel récursif ?

**Astuces pour les « Recopier et compléter » :**
- les **exemples de l'énoncé** donnent souvent le cas de base (« par convention, la factorielle de 0 est 1 ») ;
- les **lignes déjà écrites** montrent quelles variables utiliser : si l'énoncé a écrit `maj_g = majo_abs3(lst_g)`, c'est que la suite doit utiliser `maj_g` ;
- **vérifier** sa réponse avec un **petit** tableau de trace (2 ou 3 appels suffisent).

**Le cas particulier de « diviser pour régner »** (tri fusion, élément majoritaire, nombre d'inversions…) : le schéma est toujours le même :

```python
def resoudre(tab):
    if len(tab) <= 1:                 # 1. cas de base : un tableau tout petit
        return ...
    gauche = ...moitié gauche...      # 2. couper en deux
    droite = ...moitié droite...
    res_g = resoudre(gauche)          # 3. deux appels récursifs,
    res_d = resoudre(droite)          #    résultats rangés dans des variables
    return ...combiner res_g et res_d...   # 4. combiner
```

# 7. Les phrases types : « récursive » et « se termine »

Ces questions rapportent des points faciles, à condition d'être **précis**.

### « Justifier que la fonction est récursive »
> La fonction `somme_entiers_rec` est récursive car elle **s'appelle elle-même** : à la ligne 8, `return n + somme_entiers_rec(n - 1)`.

Il faut citer **la ligne** (ou l'instruction) où se trouve l'appel.

### « Expliquer pourquoi la fonction se termine »
La réponse a toujours **trois morceaux** :
> 1. à chaque appel récursif, le paramètre `n` **diminue** de 1 (ou : la chaîne perd 2 caractères, le tableau est coupé en deux…) ;
> 2. il finit donc par atteindre le **cas de base** `n == 0` (ou : une longueur inférieure à 2…) ;
> 3. dans le cas de base, il n'y a **plus d'appel récursif** : la fonction s'arrête.

### « L'appel f(…) se termine-t-il ? »
Faire le début du tableau de trace et regarder si on **atteint** le cas de base. Exemple (Métropole, septembre 2024) :

````python
def f2(n):
    if n == 0:
        return 0
    else:
        return n + f2(n - 2)


print(f2(4))   # 4 -> 2 -> 0 : on atteint le cas de base, renvoie 4 + 2 + 0 = 6
# f2(5) : 5 -> 3 -> 1 -> -1 -> -3 ... on "saute" par-dessus 0 : ne se termine pas
# (en pratique, Python s'arrête avec une RecursionError au bout d'environ 1000 appels)
````

# 8. Passer d'une fonction récursive à une boucle

Le bac demande souvent d'« écrire une version **itérative** (non récursive) ». La méthode :

| Dans la version récursive… | …devient dans la version itérative |
|:---|:---|
| la valeur du **cas de base** | la valeur **de départ** d'une variable (souvent `res`) |
| les valeurs que prend le paramètre (`n`, `n - 1`, …, `1`) | les valeurs que prend la variable de **boucle** |
| l'opération de **combinaison** (`n + …`, `n * …`) | l'opération faite **à chaque tour** sur `res` |

Exemple avec la factorielle :

````python
def fact(n):                 # récursive
    if n == 0:
        return 1             # cas de base : 1
    else:
        return n * fact(n - 1)


def fact_iteratif(n):        # itérative
    res = 1                  # le cas de base devient la valeur de départ
    for i in range(1, n + 1):
        res = res * i        # la combinaison devient l'opération de chaque tour
    return res


print(fact(5), fact_iteratif(5))
````

# 9. Fiche réflexe

| On te demande… | Tu fais… |
|:---|:---|
| pourquoi c'est **récursif** | « elle s'appelle elle-même, ligne … » |
| pourquoi ça **se termine** | paramètre qui diminue → atteint le cas de base → plus d'appel |
| ce qui est **renvoyé** | tableau de trace : descente ⬇️ puis remontée ⬆️ |
| ce qui est **affiché** | tableau de trace : `print` avant l'appel → descente ; après → remontée |
| le **nombre d'appels** | nombre de lignes du tableau, ou de cases de l'arbre (appel initial compté ?) |
| plusieurs appels dans la fonction | **arbre des appels**, gauche d'abord |
| **compléter** une fonction | les 3 questions : cas de base, plus petit problème, combiner |
| une version **itérative** | cas de base → valeur de départ, combinaison → opération dans la boucle |

**À l'écrit :** toujours **recopier** le tableau sur la copie, même si on ne le remplit pas en entier : un tableau commencé juste rapporte déjà des points.
