### Question 1.a ✍️

#### 📝 ÉNONCÉ

> Expliquer en quelques mots ce qu'est une fonction récursive.

#### ✅ RÉSOLUTION

Une fonction récursive est une fonction **qui s'appelle elle-même**. Elle contient un **cas de base**, où elle renvoie directement un résultat, et un **cas récursif**, où elle s'appelle sur un problème plus petit.

#### 💡 EXPLICATION PÉDAGOGIQUE

- Les deux mots-clés attendus : « **s'appelle elle-même** » et « **cas de base** » (ou condition d'arrêt).

### Question 1.b ✍️

#### 📝 ÉNONCÉ

> On considère la fonction Python suivante :
> ```python
> 1  def compte_rebours(n):
> 2      """ n est un entier positif ou nul """
> 3      if n >= 0:
> 4          print(n)
> 5          compte_rebours(n - 1)
> ```
> L'appel `compte_rebours(3)` affiche successivement les nombres 3, 2, 1 et 0. Expliquer pourquoi le programme s'arrête après l'affichage du nombre 0.

#### ✅ RÉSOLUTION

Après avoir affiché 0, la fonction fait l'appel `compte_rebours(-1)`. Dans cet appel, la condition `n >= 0` est **fausse** : la fonction n'affiche rien et ne fait **plus d'appel récursif**. Elle se termine, et tous les appels en attente se terminent à leur tour.

````python
def compte_rebours(n):
    """ n est un entier positif ou nul """
    if n >= 0:
        print(n)
        compte_rebours(n - 1)


compte_rebours(3)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Ici, le cas de base n'est pas écrit explicitement : c'est le cas où la condition du `if` est fausse (`n < 0`), et la fonction ne fait alors rien.
- Tableau de trace : `compte_rebours(3)` affiche 3 → `compte_rebours(2)` affiche 2 → `compte_rebours(1)` affiche 1 → `compte_rebours(0)` affiche 0 → `compte_rebours(-1)` : rien, fin.

### Question 2

#### 📝 ÉNONCÉ

> En mathématiques, la factorielle d'un entier naturel $n$ est le produit des nombres entiers strictement positifs inférieurs ou égaux à $n$. Par convention, la factorielle de 0 est 1. Par exemple :
> - la factorielle de 1 est 1 ;
> - la factorielle de 2 est 2 × 1 = 2 ;
> - la factorielle de 3 est 3 × 2 × 1 = 6 ;
> - la factorielle de 4 est 4 × 3 × 2 × 1 = 24…
>
> Recopier et compléter le programme ci-dessous afin que la fonction récursive `fact` renvoie la factorielle de l'entier passé en paramètre. Exemple : `fact(4)` renvoie 24.
> ```python
> 1  def fact(n):
> 2      """ Renvoie le produit des nombres entiers
> 3      strictement positifs inférieurs à n """
> 4      if n == 0:
> 5          return à compléter
> 6      else:
> 7          return à compléter
> ```

#### ✅ RÉSOLUTION

````python
def fact(n):
    """ Renvoie le produit des nombres entiers
    strictement positifs inférieurs à n """
    if n == 0:
        return 1
    else:
        return n * fact(n - 1)
````

*Les tests de la feuille d'exercices :*

````python
assert fact(0) == 1
assert fact(1) == 1
assert fact(4) == 24
assert fact(6) == 720
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Ligne 5 :** le cas de base est donné par l'énoncé : « par convention, la factorielle de 0 est 1 ».
- **Ligne 7 :** $n! = n \times (n-1)!$ : on multiplie `n` par le résultat de l'appel récursif.
- ⚠️ Ne pas écrire `return 0` en ligne 5 : tous les produits vaudraient 0.

| Appel | `n` | Cas de base ? | Affiche | Renvoie (en remontant) |
|:---|:---:|:---:|:---:|:---:|
| `somme_entiers_rec(3)` | 3 | non | 3 | 3 + 3 = 6 |

````python
def somme_entiers_rec(n):
    if n == 0:
        return 0
    else:
        print(n) #pour vérification
        return n + somme_entiers_rec(n - 1)


res = somme_entiers_rec(3)
print("res =", res)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

### Question 4

#### 📝 ÉNONCÉ

> Écrire en Python une fonction `somme_entiers` **non récursive** : cette fonction devra prendre en argument un entier naturel `n` et renvoyer la somme des entiers de 0 à `n` compris. Elle devra donc renvoyer le même résultat que la fonction `somme_entiers_rec` définie à la question 3. Exemple : `somme_entiers(4)` renvoie 10.

#### ✅ RÉSOLUTION

````python
def somme_entiers(n):
    s = 0
    for i in range(n + 1):
        s = s + i
    return s
````

*Les tests de la feuille d'exercices :*

````python
assert somme_entiers(0) == 0
assert somme_entiers(1) == 1
assert somme_entiers(4) == 10
assert somme_entiers(100) == 5050
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Méthode de la partie 8 du cours : le cas de base (`0`) devient la valeur de départ de `s`, et l'opération `n + ...` devient `s = s + i` à chaque tour.
- ⚠️ `range(n + 1)` et pas `range(n)` : il faut que `i` prenne aussi la valeur `n`.

>
> On a besoin de pouvoir ajouter une chaîne de caractères `s` en préfixe à chaque chaîne de caractères de la liste `liste`. On appellera cette fonction `ajouter`. Par exemple, `ajouter("a", ["b", "c"])` doit retourner `["ab", "ac"]`.
>
> **a.** Recopier et compléter le code suivant :

> def ajouter(s, liste):

>         res.................

````python
def ajouter(s, liste):
    res = []
    for m in liste:
        res.append(s + m)
    return res
````

**(c)** `ajouter("a", [""])` renvoie `["a"]` : la liste contient **une** chaîne, la chaîne vide, et `"a" + ""` vaut `"a"`.

### Question 3 ✍️

#### 📝 ÉNONCÉ

> On s'intéresse ici à la fonction suivante, où `s` est une chaîne de caractères et `n` un entier naturel.
> ```python
> def produit(s, n):
>     if n == 0:
>         return [""]
>     else:
>         res = []
>         for i in range(len(s)):
>             res = res + ajouter(s[i], produit(s, n-1))
>         return res
> ```
> **a.** Que renvoie la commande `produit("ab", 0)` ? Le résultat est-il une liste vide ?
>
> **b.** Que renvoie la commande `produit("ab", 1)` ?
>
> **c.** Que renvoie la commande `produit("ab", 2)` ?

#### ✅ RÉSOLUTION

**a.** `produit("ab", 0)` renvoie `[""]` (cas de base). Ce n'est **pas** une liste vide : elle contient un élément, la chaîne vide.

**b.** `produit("ab", 1)` : on calcule `produit("ab", 0)`, qui vaut `[""]`, puis
- pour `i = 0` : `ajouter("a", [""])` donne `["a"]` ;
- pour `i = 1` : `ajouter("b", [""])` donne `["b"]`.

Résultat : `["a", "b"]`.

**c.** `produit("ab", 2)` : on calcule `produit("ab", 1)`, qui vaut `["a", "b"]`, puis
- pour `i = 0` : `ajouter("a", ["a", "b"])` donne `["aa", "ab"]` ;
- pour `i = 1` : `ajouter("b", ["a", "b"])` donne `["ba", "bb"]`.

Résultat : `["aa", "ab", "ba", "bb"]`.

````python
def produit(s, n):
    if n == 0:
        return [""]
    else:
        res = []
        for i in range(len(s)):
            res = res + ajouter(s[i], produit(s, n-1))
        return res


print(produit("ab", 0))
print(produit("ab", 1))
print(produit("ab", 2))
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- On utilise le résultat de la question précédente à chaque fois : c'est la **remontée** du tableau de trace (on part du cas de base `n = 0` et on remonte).
- `produit(s, n)` renvoie tous les **mots de longueur `n`** qu'on peut écrire avec les lettres de `s`.

### Questions 5 à 7 — Partie B : fonction récursive ✍️

#### 📝 ÉNONCÉ

> ```python
> 1  def f2(n):
> 2      if n == 0:
> 3          return 0
> 4      else:
> 5          return n + f2(n-2)
> ```
> 5. L'appel `f2(4)` termine-t-il ? Si oui, indiquer la valeur renvoyée par `f2(4)` ; sinon, justifier brièvement.
> 6. L'appel `f2(5)` termine-t-il ? Si oui, indiquer la valeur renvoyée par `f2(5)` ; sinon, justifier brièvement.
> 7. Déterminer l'ensemble des entiers naturels `n` pour lesquels l'appel `f2(n)` termine.

#### ✅ RÉSOLUTION

5. Oui. Tableau de trace :

| Appel | `n` | Cas de base ? | Renvoie (en remontant) |
|:---|:---:|:---:|:---:|
| `f2(4)` | 4 | non | 4 + 2 = 6 |
| `f2(2)` | 2 | non | 2 + 0 = 2 |
| `f2(0)` | 0 | oui | 0 |

`f2(4)` termine et renvoie **6**.

6. Non. Les appels successifs sont `f2(5)`, `f2(3)`, `f2(1)`, `f2(-1)`, `f2(-3)`… Le paramètre diminue de 2 à chaque fois et « saute » par-dessus 0 : on n'atteint **jamais** le cas de base `n == 0`.
7. `f2(n)` termine pour tous les entiers naturels **pairs**.

````python
def f2(n):
    if n == 0:
        return 0
    else:
        return n + f2(n-2)


print(f2(4))
try:
    f2(5)
except RecursionError:
    print("f2(5) : RecursionError, l'appel ne termine pas")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- C'est le contre-exemple parfait de la phrase type « le paramètre diminue, donc il atteint le cas de base » : ici il diminue, mais il **n'atteint pas** le cas de base quand `n` est impair.
- En pratique, Python s'arrête avec une `RecursionError` au bout d'environ 1000 appels : ce n'est pas une valeur renvoyée, c'est une erreur.

### Question 1 ✍️

#### 📝 ÉNONCÉ

> **(a)** Choisir parmi les propositions ci-dessous celle qui convient pour compléter la fonction `palindrome` (lignes 3 et 4) :
>
> | Proposition 1 | Proposition 2 | Proposition 3 | Proposition 4 |
> |:---|:---|:---|:---|
> | `if len(txt) <= 2:`<br>`    return True` | `if len(txt) < 2:`<br>`    return False` | `if len(txt) < 2:`<br>`    return True` | `if len(txt) <= 2:`<br>`    return False` |

#### ✅ RÉSOLUTION

**(a)** La **proposition 3** : un mot de 0 ou 1 lettre est un palindrome, il faut donc renvoyer `True` quand `len(txt) < 2`.
- Les propositions 2 et 4 renvoient `False` pour `""` ou `"A"`, qui sont des palindromes.
- La proposition 1 renvoie `True` pour `"AB"`, qui n'est pas un palindrome.

**(b)** `txt[0]` vaut `"b"`, `txt[taille - 1]` vaut `"r"` (`taille` vaut 7), et `interieur` vaut `"onjou"` (les lettres d'indices 1 à 5).

````python
def palindrome(txt):
    """ Str -> Bool """
    if len(txt) < 2:
        return True
    taille = len(txt)
    interieur = ""
    for i in range(1, taille - 1):
        interieur = interieur + txt[i]
    return (txt[0] == txt[taille - 1]) and (palindrome(interieur))


for mot in ["", "A", "AB", "BOB", "LAVAL", "bonjour"]:
    print(repr(mot), palindrome(mot))
````

## Exercice 8 — Le tri fusion (1)  ⭐⭐

📚 **Tombé au bac :** épreuve écrite, **Métropole 2021, sujet 1**, exercice 4

*Cet exercice porte sur l'algorithme de tri fusion, qui s'appuie sur la méthode dite de « diviser pour régner ».*

L'algorithme de tri fusion utilise deux fonctions `moitie_gauche` et `moitie_droite` qui prennent en argument une liste `L` et renvoient respectivement :
- la sous-liste de `L` formée des éléments d'indice strictement inférieur à `len(L)//2` ;
- la sous-liste de `L` formée des éléments d'indice supérieur ou égal à `len(L)//2`.

On rappelle que la syntaxe `a//b` désigne la division entière de `a` par `b`. Par exemple :
```python
>>> L = [3, 5, 2, 7, 1, 9, 0]            >>> M = [4, 1, 11, 7]
>>> moitie_gauche(L)                     >>> moitie_gauche(M)
[3, 5, 2]                                [4, 1]
>>> moitie_droite(L)                     >>> moitie_droite(M)
[7, 1, 9, 0]                             [11, 7]
```
L'algorithme utilise aussi une fonction `fusion` qui prend en argument deux listes triées `L1` et `L2` et renvoie une liste `L` triée et composée des éléments de `L1` et `L2`.

On donne ci-dessous le code Python d'une fonction récursive `tri_fusion` qui prend en argument une liste `L` et renvoie une nouvelle liste triée formée des éléments de `L`.
```python
def tri_fusion(L):
    n = len(L)
    if n<=1 :
        return L
    print(L)
    mg = moitie_gauche(L)
    md = moitie_droite(L)
    L1 = tri_fusion(mg)
    L2 = tri_fusion(md)
    return fusion(L1, L2)
```

### Question 1 ✍️

#### 📝 ÉNONCÉ

> **a.** Quel est l'ordre de grandeur du coût, en nombre de comparaisons, de l'algorithme de tri fusion pour une liste de longueur $n$ ?
>
> **b.** Citer le nom d'un autre algorithme de tri. Donner l'ordre de grandeur de son coût, en nombre de comparaisons, pour une liste de longueur $n$. Comparer ce coût à celui du tri fusion. Aucune justification n'est attendue.

#### ✅ RÉSOLUTION

**a.** Le tri fusion a un coût de l'ordre de $n \log_2 n$ (coût **quasi-linéaire**).

**b.** Le **tri par insertion** (ou le tri par sélection) a un coût de l'ordre de $n^2$ (coût **quadratique**) dans le pire des cas. C'est beaucoup plus que $n \log_2 n$ : pour $n = 1\,000\,000$, environ $10^{12}$ comparaisons contre environ $2 \times 10^7$.

#### 💡 EXPLICATION PÉDAGOGIQUE

- À connaître par cœur : tri par insertion / sélection → $n^2$ ; tri fusion → $n \log_2 n$.

> Dans cette fonction, les entiers `i1` et `i2` représentent respectivement les indices des éléments des listes `L1` et `L2` que l'on souhaite comparer :
> - si aucun des deux indices n'est valide, la boucle `while` est interrompue ;
> - si `i1` n'est plus un indice valide, on va ajouter à `L` les éléments de `L2` à partir de l'indice `i2` ;
> - si `i2` n'est plus un indice valide, on va ajouter à `L` les éléments de `L1` à partir de l'indice `i1` ;
> - sinon, le plus petit élément non encore traité est ajouté à `L` et on décale l'indice correspondant.
>
> Écrire les instructions manquantes des lignes 17 à 22 permettant d'insérer dans la liste `L` les éléments des listes `L1` et `L2` par ordre croissant.

#### ✅ RÉSOLUTION

````python
def fusion(L1, L2):
    L = []
    n1 = len(L1)
    n2 = len(L2)
    i1 = 0
    i2 = 0
    while i1 < n1 or i2 < n2 :
        if i1 >= n1:
            L.append(L2[i2])
            i2 = i2 + 1
        elif i2 >= n2:
            L.append(L1[i1])
            i1 = i1 + 1
        else:
            e1 = L1[i1]
            e2 = L2[i2]
            if e1 <= e2:
                L.append(e1)
                i1 = i1 + 1
            else:
                L.append(e2)
                i2 = i2 + 1
    return L
````

## Exercice 10 — Le tri de Stooge  ⭐⭐⭐

📚 **Tombé au bac :** épreuve écrite, **Amérique du Nord 2024, jour 2**, exercice 1

*Cet exercice porte sur les algorithmes de tri et la programmation en Python.*

On se propose dans cet exercice de se pencher sur un algorithme pour trier un tableau appelé le **tri de Stooge**.

Pour trier les éléments situés entre les indices `i` et `j`, où `i < j`, dans un tableau `t` par ce tri, on procède ainsi :
- si les éléments d'indice `i` et `j` sont mal placés, on les échange ;
- s'il y a au moins trois éléments entre les indices `i` et `j` :
  - on trie les deux premiers tiers du tableau avec cette méthode ;
  - on trie les deux derniers tiers du tableau avec cette méthode ;
  - on trie à nouveau les deux premiers tiers du tableau avec cette méthode.

Pour réaliser ce découpage en tiers, on considère l'entier `k` défini par l'expression `(j - i + 1) // 3`, et on considère les indices intermédiaires `i+k` et `j-k` :
```
         i          i+k              j-k          j
   t   [ .......... | ............... | .......... ]
       <------ deux premiers tiers ---->
                    <------ deux derniers tiers -->
```
Voici le code partiel de l'algorithme du tri de Stooge en Python, qui trie les éléments d'un tableau par ordre croissant.
```python
1  def triStooge(tab, i, j):
2      if tab[i] > tab[j]:
3          echange(tab, i, j)
4      if (j - i) > 1:
5          k = (j - i + 1)//3
6          triStooge(...)
7          triStooge(...)
8          triStooge(...)
```

### Questions 1 et 2

#### 📝 ÉNONCÉ

> 1. Écrire la fonction `echange(tab, i, j)` qui prend en arguments une liste Python `tab` et deux indices `i`, `j`, et réalise sur place l'échange des valeurs dans `tab` à ces indices. La fonction ne renvoie rien.
> 2. Finaliser le programme précédent en complétant les lignes 6, 7 et 8.

#### ✅ RÉSOLUTION

````python
def echange(tab, i, j):
    temp = tab[i]
    tab[i] = tab[j]
    tab[j] = temp


def triStooge(tab, i, j):
    if tab[i] > tab[j]:
        echange(tab, i, j)
    if (j - i) > 1:
        k = (j - i + 1)//3
        triStooge(tab, i, j - k)
        triStooge(tab, i + k, j)
        triStooge(tab, i, j - k)
````

## Exercice 11 — L'élément majoritaire  ⭐⭐⭐

Dans cet exercice, on s'intéresse à des algorithmes pour déterminer, s'il existe, l'élément **absolument majoritaire** d'une liste. On dit qu'un élément est absolument majoritaire s'il apparaît dans **strictement plus de la moitié** des emplacements de la liste.

Par exemple, la liste `[1, 4, 1, 6, 1, 7, 2, 1, 1]` admet 1 comme élément absolument majoritaire, car il apparaît 5 fois sur 9 éléments. Par ailleurs, la liste `[1, 4, 6, 1, 7, 2, 1, 1]` n'admet pas d'élément absolument majoritaire, car celui qui est le plus fréquent est 1, mais il n'apparaît que 4 fois sur 8, ce qui ne fait pas plus que la moitié.

### Question 1 ✍️

#### 📝 ÉNONCÉ

> Déterminer les effectifs possibles d'un élément absolument majoritaire dans une liste de taille 10.

#### ✅ RÉSOLUTION

Il doit apparaître strictement plus de 10 / 2 = 5 fois : ses effectifs possibles sont **6, 7, 8, 9 ou 10**.

#### 💡 EXPLICATION PÉDAGOGIQUE

- « Strictement plus de la moitié » : 5 sur 10 ne suffit pas.

### Question 2

#### 📝 ÉNONCÉ

#### ✅ RÉSOLUTION

````python
def effectif(val, lst):
    compteur = 0
    for element in lst:
        if element == val:
            compteur = compteur + 1
    return compteur
````

## Exercice 12 — Compter les inversions  ⭐⭐⭐

*Cet exercice traite de manipulation de tableaux, de récursivité et du paradigme « diviser pour régner ».*

Dans un tableau Python d'entiers `tab`, on dit que le couple d'indices $(i, j)$ forme une **inversion** lorsque $i < j$ et `tab[i] > tab[j]`. Quelques exemples :

On se propose dans cet exercice de déterminer le nombre d'inversions dans un tableau quelconque.

**b.** `fonction1(tab, i)` renvoie le **nombre d'inversions dont le premier indice est `i`**, c'est-à-dire le nombre d'éléments situés après l'indice `i` et strictement plus petits que `tab[i]`.

### Question A.2

#### 📝 ÉNONCÉ

> En utilisant la fonction précédente, écrire une fonction `nombre_inversions(tab)` qui prend en argument un tableau et renvoie le nombre d'inversions dans ce tableau.
> ```python
> >>> nombre_inversions([1, 5, 7])
> 0
> >>> nombre_inversions([1, 6, 2, 7, 3])
> 3
> >>> nombre_inversions([7, 6, 5, 3])
> 6
> ```

#### ✅ RÉSOLUTION

````python
def fonction1(tab, i):
    nb_elem = len(tab)
    cpt = 0
    for j in range(i+1, nb_elem):
        if tab[j] < tab[i]:
            cpt += 1
    return cpt


def nombre_inversions(tab):
    total = 0
    for i in range(len(tab)):
        total = total + fonction1(tab, i)
    return total
````

*Les tests de la feuille d'exercices :*

#### 💡 EXPLICATION PÉDAGOGIQUE

- Chaque inversion $(i, j)$ est comptée exactement une fois, par `fonction1(tab, i)`. Il suffit donc d'additionner pour tous les `i`.
- Coût : deux boucles imbriquées, donc de l'ordre de $n^2$ (quadratique).

### Question B.3

#### 📝 ÉNONCÉ

> Dans la suite, on suppose qu'on dispose de la fonction `moitie_droite(tab)` qui renvoie la moitié droite sans l'élément du milieu.
>
> On suppose qu'une fonction `nb_inv_tab(tab1, tab2)` a été écrite. Cette fonction renvoie le nombre d'inversions du tableau obtenu en mettant bout à bout les tableaux `tab1` et `tab2`, à condition que `tab1` et `tab2` soient triés dans l'ordre croissant.
> ```python
> >>> nb_inv_tab([3, 7, 9], [2, 10])
> 3
> >>> nb_inv_tab([7, 9, 13], [7, 10, 14])
> 3
> ```
> En utilisant la fonction `nb_inv_tab` et les questions précédentes, écrire une fonction récursive `nb_inversions_rec(tab)` qui permet de calculer le nombre d'inversions dans un tableau. Cette fonction renverra le même nombre que `nombre_inversions(tab)` de la partie A. On procédera de la façon suivante :
> - séparer le tableau en deux tableaux de tailles égales (à une unité près) ;
> - appeler récursivement la fonction `nb_inversions_rec` pour compter le nombre d'inversions dans chacun des deux tableaux ;
> - trier les deux tableaux (on rappelle qu'une fonction de tri est déjà définie) ;
> - ajouter au nombre d'inversions précédemment comptées le nombre renvoyé par la fonction `nb_inv_tab` avec pour arguments les deux tableaux triés.
>
> *(Les fonctions `tri`, `moitie_droite` et `nb_inv_tab` sont fournies dans la cellule ci-dessous.)*

#### ✅ RÉSOLUTION

````python
# Fonctions fournies par le sujet (ne pas modifier)
def tri(tab):
    res = []
    for x in tab:
        res.append(x)
    for i in range(1, len(res)):
        x = res[i]
        j = i - 1
        while j >= 0 and res[j] > x:
            res[j + 1] = res[j]
            j = j - 1
        res[j + 1] = x
    return res


def moitie_droite(tab):
    n = len(tab)
    res = []
    for i in range((n + 1) // 2, n):
        res.append(tab[i])
    return res


def nb_inv_tab(tab1, tab2):
    cpt = 0
    for x in tab1:
        for y in tab2:
            if x > y:
                cpt = cpt + 1
    return cpt


def nb_inversions_rec(tab):
    if len(tab) <= 1:
        return 0
    gauche = moitie_gauche(tab)
    droite = moitie_droite(tab)
    inv_g = nb_inversions_rec(gauche)
    inv_d = nb_inversions_rec(droite)
    return inv_g + inv_d + nb_inv_tab(tri(gauche), tri(droite))
````

*Les tests de la feuille d'exercices :*

````python
assert nb_inv_tab([3, 7, 9], [2, 10]) == 3
assert nb_inv_tab([7, 9, 13], [7, 10, 14]) == 3
assert nb_inversions_rec([1, 5, 7]) == 0
assert nb_inversions_rec([1, 6, 2, 7, 3]) == 3
assert nb_inversions_rec([7, 6, 5, 3]) == 6
assert nb_inversions_rec([]) == 0
for t in [[4, 8, 3, 7], [9, 1, 8, 2, 7, 3], [5, 5, 1]]:
    assert nb_inversions_rec(t) == nombre_inversions(t)
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **Cas de base :** un tableau de 0 ou 1 élément n'a **aucune** inversion. L'énoncé ne le donne pas : il faut y penser soi-même (sinon la récursion ne s'arrête jamais).
- Une inversion $(i, j)$ est de l'un des trois types : $i$ et $j$ tous les deux à gauche (comptée par `inv_g`), tous les deux à droite (`inv_d`), ou $i$ à gauche et $j$ à droite (comptée par `nb_inv_tab`). Trier chaque moitié ne change pas ce troisième nombre.
- C'est encore le schéma « diviser pour régner » de la partie 6 du cours.

> ```

#### ✅ RÉSOLUTION

````python
class Marchandise:
    def __init__(self, p, v):
        assert v > 0
        self.prix = p
        self.volume = v


def prixListe(tab):
    total = 0
    for m in tab:
        total = total + m.prix
    return total


def chargeOptimale(tab, v_restant, i):
    if i>=len(tab):
        return []
    else:
        if tab[i].volume > v_restant:
            return chargeOptimale(tab, v_restant, i+1)
        else:
            option1 = chargeOptimale(tab, v_restant, i+1)
            option2 = [tab[i]] + chargeOptimale(tab, v_restant - tab[i].volume, i+1)
            if prixListe(option1) > prixListe(option2):
                return option1
            else:
                return option2
````

*Les tests de la feuille d'exercices :*

````python
m1 = Marchandise(40, 20)
m2 = Marchandise(210, 70)
m3 = Marchandise(160, 40)
m4 = Marchandise(50, 50)
tab = [m3, m2, m1, m4]
charge = chargeOptimale(tab, 100, 0)
assert prixListe(charge) == 250
assert m1 in charge and m2 in charge
assert chargeOptimale(tab, 10, 0) == []
assert prixListe(chargeOptimale(tab, 1000, 0)) == 460
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- ⚠️ Dans le code du sujet, `option1` est l'option **sans** la marchandise `i`, et `option2` celle **avec** (on le voit au `[tab[i]] + …`). C'est l'inverse de l'ordre du texte : il faut lire le code.
- Si on prend la marchandise `i`, le volume restant **diminue** de son volume : `v_restant - tab[i].volume`.
- Dans les deux options, on passe à la marchandise suivante : `i+1`. C'est ce qui garantit que la fonction se termine (`i` augmente jusqu'à `len(tab)`).
- Chaque appel en fait deux : le nombre d'appels **double** à chaque marchandise ($2^n$ en tout), d'où un coût exponentiel. La programmation dynamique permet de faire mieux (exercice suivant).

### Question 6

#### 📝 ÉNONCÉ

> On dira qu'un conduit est **maximal** si son score de confiance est maximal. On observe que :
> - si on a des conduits maximaux pour les deux petites pyramides obtenues en enlevant le sommet, on obtient un conduit maximal en ajoutant le sommet devant le conduit de plus grand score parmi les deux. Par exemple, pour `ex1`, la petite pyramide de gauche (de sommet 6) a un conduit maximal de score 6 + 5 + 6 + 5 = 22, celle de droite (de sommet 2) un conduit maximal de score 2 + 7 + 6 + 5 = 20 : le score maximal pour `ex1` est donc 4 + 22 = 26 ;
> - si la pyramide n'a qu'un seul niveau, le seul conduit possible contient le sommet, et le nombre de ce sommet est le score maximal.
>
> On note `score_max(i, j, p)` le score maximal possible depuis le nombre d'indice `j` du niveau `i` de la pyramide `p`. On a alors les relations suivantes :
> - `score_max(len(p)-1, j, p) = p[len(p)-1][j]` ;
> - `score_max(i, j, p) = p[i][j] + max(score_max(i+1, j, p), score_max(i+1, j+1, p))`.
>
> Le score maximal possible pour `p` toute entière sera alors `score_max(0, 0, p)`.
>
> Écrire la fonction récursive `score_max` qui implémente les règles précédentes.

#### ✅ RÉSOLUTION

````python
def score_max(i, j, p):
    if i == len(p) - 1:
        return p[i][j]
    return p[i][j] + max(score_max(i + 1, j, p), score_max(i + 1, j + 1, p))
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Les deux relations de l'énoncé se traduisent **mot pour mot** : la première est le cas de base (dernier niveau), la deuxième le cas récursif.
- Deux appels récursifs par appel : comme pour Fibonacci, on recalcule plusieurs fois les mêmes cases (par exemple `score_max(3, 1, p)` est calculé depuis `score_max(2, 0, p)` **et** depuis `score_max(2, 1, p)`).

### Questions 7 et 8 — programmation dynamique

#### 📝 ÉNONCÉ

> Pour éviter la redondance des calculs, on construit une pyramide `s` dont le nombre à l'indice `j` du niveau `i` correspond à `score_max(i, j, p)`.
>
> 7. Écrire une fonction `pyramide_nulle` qui prend en paramètre un entier `n` et construit une pyramide remplie de 0 à `n` niveaux.
> 8. Compléter la fonction `prog_dyn` ci-dessous qui prend en paramètre une pyramide `p`, et qui renvoie le score maximal pour un conduit dans `p`. Pour cela, on construit une pyramide `s` remplie de 0 de la même taille et on la remplit avec les valeurs de `score_max` en commençant par le dernier niveau et en appliquant petit à petit les relations données ci-dessus.
> ```python
> 1  def prog_dyn(p):
> 2      n = len(p)
> 3      s = ...
> 4      # remplissage du dernier niveau
> 5      for j in ...
> 6          s[n-1][j] = ...
> 7      # remplissage des autres niveaux
> 8      for i in ...
> 9          for j in ...
> 10             s[i][j] = ...
> 11     # renvoie du score maximal
> 12     return s[0][0]
> ```

#### ✅ RÉSOLUTION

````python
def pyramide_nulle(n):
    p = []
    for i in range(n):
        niveau = []
        for j in range(i + 1):
            niveau.append(0)
        p.append(niveau)
    return p


def prog_dyn(p):
    n = len(p)
    s = pyramide_nulle(n)
    # remplissage du dernier niveau
    for j in range(n):
        s[n-1][j] = p[n-1][j]
    # remplissage des autres niveaux
    for i in range(n - 2, -1, -1):
        for j in range(i + 1):
            s[i][j] = p[i][j] + max(s[i+1][j], s[i+1][j+1])
    # renvoie du score maximal
    return s[0][0]
````

*Les tests de la feuille d'exercices :*

````python
assert pyramide_nulle(1) == [[0]]
assert pyramide_nulle(3) == [[0], [0, 0], [0, 0, 0]]
ex1 = [[4], [6, 2], [3, 5, 7], [5, 1, 6, 2], [4, 7, 3, 5, 2]]
ex2 = [[3], [1, 2], [4, 5, 9], [3, 6, 2, 1]]
assert prog_dyn(ex1) == 26
assert prog_dyn(ex2) == 16
assert prog_dyn([[7]]) == 7
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- **`pyramide_nulle` :** le niveau `i` contient `i + 1` cases. On construit chaque niveau avec une boucle, puis on l'ajoute à la pyramide.
- **`prog_dyn` :** on remplit `s` **du bas vers le haut** (`range(n - 2, -1, -1)` donne `n - 2`, …, `0`). Quand on calcule `s[i][j]`, les deux cases du dessous sont déjà connues : plus besoin d'appel récursif, chaque case est calculée **une seule fois**.
- C'est la **remontée** du tableau de trace, faite directement avec des boucles.

### Questions 9 et 10 ✍️

#### 📝 ÉNONCÉ

> 9. Montrer que le coût d'exécution de la fonction `prog_dyn` est quadratique en `n` pour une pyramide à `n` niveaux.
> 10. Expliquer comment adapter la fonction `score_max` pour éviter la redondance des calculs afin d'obtenir également un coût quadratique, tout en gardant une approche récursive.

#### ✅ RÉSOLUTION

9. `prog_dyn` calcule chaque case de la pyramide **une seule fois**, en temps constant. La pyramide contient $1 + 2 + \cdots + n = \dfrac{n(n+1)}{2}$ cases : le coût est de l'ordre de $n^2$, donc **quadratique**.

10. On utilise la **mémoïsation** : on garde en mémoire (par exemple dans un dictionnaire, dont les clés sont les couples `(i, j)`) les valeurs de `score_max` déjà calculées. Au début de chaque appel, si la valeur de `(i, j)` est déjà dans le dictionnaire, on la renvoie directement sans refaire d'appel récursif ; sinon, on la calcule, on la range dans le dictionnaire, puis on la renvoie. Chaque case n'est alors calculée qu'une seule fois.

````python
# Version récursive avec mémoïsation (question 10)
def score_max_memo(i, j, p, memo):
    if (i, j) in memo:
        return memo[(i, j)]
    if i == len(p) - 1:
        res = p[i][j]
    else:
        res = p[i][j] + max(score_max_memo(i + 1, j, p, memo), score_max_memo(i + 1, j + 1, p, memo))
    memo[(i, j)] = res
    return res


ex1 = [[4], [6, 2], [3, 5, 7], [5, 1, 6, 2], [4, 7, 3, 5, 2]]
print(score_max_memo(0, 0, ex1, {}))
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- La mémoïsation, c'est « ne jamais refaire deux fois le même calcul ». On l'a déjà vue avec Fibonacci dans la feuille principale.
